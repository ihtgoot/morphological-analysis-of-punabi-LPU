```python
dataset = "/home/ihtgoot/college_project/NLP_project/NEW/cleaned_shloks.csv"
df = pd.read_csv(dataset,index_col=0)
df["shlok_text"]
```

```python
all_lines = df["shlok_text"].astype(str).tolist()
vocab = set()

for line in all_lines:
    for w in line.split():
        vocab.add(w)

vocab = list(vocab)
print("vocab size : ", len(vocab))
```

```python
#  normalize text 
from indicnlp.normalize.indic_normalize import IndicNormalizerFactory


def normalize_word(w):
    normalizer = IndicNormalizerFactory().get_normalizer("pa")
    return normalizer.normalize(w)

norm_vocab = [normalize_word(w) for w in vocab]
```

```python
# Longest Common Prefix, to find stem
def lcp(a,b):
    m = min(len(a),len(b))
    for i in range(m):
        if a[i]!=b[i]:
            return a[:i]
    return a[:m]
```

```python
from collections import Counter

def build_stems(vocab, min_prefix_len=3, min_family=4) :
    stems = Counter()
    vocab_sorted = sorted(vocab)
    for i in range(len(vocab_sorted)-1):
          w1 = vocab_sorted[i]
          w2 = vocab_sorted[i+1]
          LPC = lcp(w1,w2)
          if len(LPC)>= min_prefix_len:
            stems[LPC]+=1
    return {s for s,c in stems.items() if c >= min_family}

stems = build_stems(norm_vocab)
print("Stem count:", len(stems))
```

```python
def build_suffixes(vocab, stems, min_freq=4):
    suf = Counter()
    for w in vocab:
        for s in stems:
            if w.startswith(s):
                rem = w[len(s):]
                if 1 <= len(rem) <= 3:
                    suf[rem] += 1
    return {s for s,c in suf.items() if c >= min_freq}

suffixes = build_suffixes(norm_vocab, stems)
print("Suffix count:", len(suffixes))
```

```python
# WORD = STEM + SUFFIX
# Similarity = Levenshtein edit distance
# Lower distance = better match.

from Levenshtein import distance

def find_stem(word, stem, max_edit):
    best_stem = None
    best_score = 500

    for s in stems:
        if len(s) > len(word):
            continue
        d = distance(word[:len(s)],s)

        if d < best_score :
            best_score = d
            best_stem = s
    
    if best_score >max_edit:
        return None

    return best_stem
```

```python
def remove_stem(word,stem):
    if word.startswith(stem):
        return word[len(stem):]
    else:
        idx = word.find(stem)
        return word[idx+len(stem):]
    
```

```python
# identify suffix

def find_suffix(remainder, suffixes):
    if remainder in suffixes:
        return remainder

    for i in range(1, len(remainder)+1):
        suf = remainder[:i]
        if suf in suffixes:
            return suf

    return None
```

```python
def segment_word(word, stems, suffixes):
    stem = find_stem(word, stems,1)
    if stem is None:
        return None

    remainder = remove_stem(word, stem)

    if remainder == "":
        return (stem, None)

    suffix = find_suffix(remainder, suffixes)

    return (stem, suffix)
```
