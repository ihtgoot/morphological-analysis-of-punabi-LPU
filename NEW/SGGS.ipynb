```python
import pdfplumber

input_pdf = "Guru_Granth_Sahib_Ji.pdf"       
output_txt = "SGGS_G.txt"

with pdfplumber.open(input_pdf) as pdf, open(output_txt, "w", encoding="utf-8") as out:
    for i, page in enumerate(pdf.pages):
        text = page.extract_text()
        if text:
            out.write(text)

print("Done. Extracted to:", output_txt)
```
